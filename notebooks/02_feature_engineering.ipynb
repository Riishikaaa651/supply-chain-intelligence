```python
import pandas as pd
import numpy as np
# load the cleaned dataset
df = pd.read_csv(r"C:\Users\Rishj\OneDrive\Desktop\PROJECTS\supply_chain_intelligence\data\processed\eda_cleaned_dataset.csv")
df.shape
print(df.columns.tolist())
```

```python
#see how consistent the scheduled delivery time of the shipping mode is
shipping_mode_stats = df.groupby('Shipping Mode')['Days for shipment (scheduled)'].agg(['mean', 'std'])
shipping_mode_stats.columns = ['scheduled mean', 'scheduled std']
print(shipping_mode_stats)
```

```python
#calculating the gap between thhe actual and scheduled shipping days
df['shipping_gap'] = df['Days for shipping (real)'] - df['Days for shipment (scheduled)']

# Group by supplier — check what column represents "supplier" in your dataset
# (DataCo doesn't have an explicit "Supplier" column — it's often approximated by'Department Name', 'Category Name', or you may need to treat Product/Category as a proxy)
supplier_stats = df.groupby('Department Name')['shipping_gap'].agg(['mean', 'std', 'count'])
supplier_stats.columns = ['avg_delay_gap', 'delay_gap_std', 'order_count']

print(supplier_stats.sort_values('delay_gap_std', ascending=False))
```

```python
print(df.columns.tolist())
```

```python
category_stats = df.groupby('Category Name')['shipping_gap'].agg(['mean', 'std', 'count'])
category_stats.columns = ['avg_delay_gap', 'delay_gap_std', 'order_count']
print(category_stats.sort_values('delay_gap_std', ascending=False))
```

```python
#order region analysis
region_stats = df.groupby('Order Region')['shipping_gap'].agg(['mean', 'std', 'count'])
region_stats.columns = ['avg_delay_gap', 'delay_gap_std', 'order_count']
print(region_stats.sort_values('delay_gap_std', ascending=False))
```

```python
#market column analysis
market_stats = df.groupby('Market')['shipping_gap'].agg(['mean', 'std', 'count'])
market_stats.columns = ['avg_delay_gap', 'delay_gap_std', 'order_count']
print(market_stats.sort_values('delay_gap_std', ascending=False))
```

```python
#shipping mode analysis 
shipping_mode_gap_stats = df.groupby('Shipping Mode')['shipping_gap'].agg(['mean', 'std', 'count'])
shipping_mode_gap_stats.columns = ['avg_delay_gap', 'delay_gap_std', 'order_count']
print(shipping_mode_gap_stats.sort_values('delay_gap_std', ascending=False))
```

```python
#shipping mode reliability features 
# Map the mean and std you already calculated back onto each row
mode_stats = df.groupby('Shipping Mode')['shipping_gap'].agg(['mean', 'std'])
mode_stats.columns = ['mode_avg_delay', 'mode_delay_std']
df = df.merge(mode_stats, on='Shipping Mode', how='left')
```

```python
#order region reliability features
region_stats = df.groupby('Order Region')['shipping_gap'].agg(['mean', 'std'])
region_stats.columns = ['region_avg_delay', 'region_delay_std']
df = df.merge(region_stats, on='Order Region', how='left')
```

```python
#2: seasonal/time-based features
# Extract calendar signals from order date
df['order date (DateOrders)'] = pd.to_datetime(df['order date (DateOrders)'])
df['order_month'] = df['order date (DateOrders)'].dt.month
df['order_day_of_week'] = df['order date (DateOrders)'].dt.dayofweek  # 0=Monday, 6=Sunday
df['order_quarter'] = df['order date (DateOrders)'].dt.quarter
df['order_year'] = df['order date (DateOrders)'].dt.year

# Optional: holiday season flag (Nov-Dec, common peak-order period)
df['is_holiday_season'] = df['order_month'].isin([11, 12]).astype(int)
```

```python
monthly_stats = df.groupby('order_month')['shipping_gap'].agg(['mean', 'std', 'count'])
monthly_stats.columns = ['avg_delay_gap', 'delay_gap_std', 'order_count']

print(monthly_stats.sort_values('delay_gap_std', ascending=False))
```

```python
#order day of week analysis
dow_stats = df.groupby('order_day_of_week')['shipping_gap'].agg(['mean', 'std', 'count'])
dow_stats.columns = ['avg_delay_gap', 'delay_gap_std', 'order_count']
print(dow_stats.sort_values('delay_gap_std', ascending=False))
```

```python
#order_quarter and order_year
quarter_stats = df.groupby('order_quarter')['shipping_gap'].agg(['mean', 'std', 'count'])
quarter_stats.columns = ['avg_delay_gap', 'delay_gap_std', 'order_count']
print(quarter_stats.sort_values('delay_gap_std', ascending=False))
print()
year_stats = df.groupby('order_year')['shipping_gap'].agg(['mean', 'std', 'count'])
year_stats.columns = ['avg_delay_gap', 'delay_gap_std', 'order_count']
print(year_stats.sort_values('delay_gap_std', ascending=False))
```

```python
#3: Supplier historical reliability score — past order's % late/avg delay (leakage-safe, sirf Pehle ke data se)
# Step 1: Make sure data is sorted by order date (so "before" actually means before)
df = df.sort_values('order date (DateOrders)').reset_index(drop=True)

# Step 2: Calculate expanding (cumulative, before-only) mean of shipping_gap per Shipping Mode
df['mode_reliability_score'] = (
    df.groupby('Shipping Mode')['shipping_gap']
      .transform(lambda x: x.expanding().mean().shift(1))
)
```

```python
# Check how many NaNs you get
print(df['mode_reliability_score'].isnull().sum())

# Fill NaNs with overall average as a fallback
df['mode_reliability_score'] = df['mode_reliability_score'].fillna(df['shipping_gap'].mean())
```

```python
df['mode_reliability_score'] = df['mode_reliability_score'].fillna(df['shipping_gap'].mean())

# Confirm no NaNs remain
print(df['mode_reliability_score'].isnull().sum())
```

```python
#order region historical reliability score
df['region_reliability_score'] = (
    df.groupby('Order Region')['shipping_gap']
      .transform(lambda x: x.expanding().mean().shift(1))
)

# Check NaN count (expect roughly one per Order Region group)
print(df['region_reliability_score'].isnull().sum())
```

```python
df['region_reliability_score'] = df['region_reliability_score'].fillna(df['shipping_gap'].mean())
print(df['region_reliability_score'].isnull().sum())  # should be 0
```

```python
#4: categorical encoding
# One-hot encode Shipping Mode (only 4 categories - safe to expand)
df = pd.get_dummies(df, columns=['Shipping Mode'], prefix='mode')

# Order Region: since you already have region_avg_delay and region_reliability_score 
# capturing its signal numerically, you likely don't need to one-hot encode it separately.
# But if you still want the raw category available too, check its cardinality first:
print(df['Order Region'].nunique())
```
